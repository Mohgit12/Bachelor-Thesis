```python
import geopandas as gpd

# Load municipalities layer
gpkg_path = r"c:\Users\douaa\Downloads\WijkBuurtkaart_2025_v1\WijkBuurtkaart_2025_v1\wijkenbuurten_2025_v1.gpkg"
gemeenten = gpd.read_file(gpkg_path, layer="gemeenten")

# Filter for Amsterdam
amsterdam = gemeenten[gemeenten["gemeentenaam"] == "Amsterdam"]

# Check the result
print(amsterdam)
 
```
